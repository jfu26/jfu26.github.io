# Expectations and Fixed Points in a Linear Model

**Audience.** Beginning graduate students in macroeconomics.

**Prerequisites.** Linear equations, conditional expectations, and basic Python.

**Learning goal.** Solve a forward-looking equilibrium, verify the stability condition, and study comparative statics.

## Model

Consider the equilibrium condition

$$x_t = a + \beta \mathbb{E}_t[x_{t+1}].$$

In a stationary deterministic equilibrium, $x_t=x^*$, so the fixed point is $x^*=a/(1-\beta)$ whenever $|\beta|<1$.

```python
from __future__ import annotations

a = 1.0
beta = 0.85
assert abs(beta) < 1, "A stable fixed point requires |beta| < 1."
x_star = a / (1 - beta)
round(x_star, 3)
```

## Iteration

Starting away from equilibrium, repeatedly apply $T(x)=a+\beta x$. Convergence is geometric because the slope of the mapping is $\beta$.

```python
path = [0.0]
for _ in range(12):
    path.append(a + beta * path[-1])

[(period, round(value, 3)) for period, value in enumerate(path)]
```

## Comparative statics

The multiplier $\partial x^*/\partial a=1/(1-\beta)$ grows as expectations become more persistent.

```python
betas = [0.0, 0.25, 0.5, 0.75, 0.9]
comparative_statics = [(b, round(a / (1 - b), 2)) for b in betas]
comparative_statics
```
